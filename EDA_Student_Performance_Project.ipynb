```python
# Project 3: Exploratory Data Analysis (EDA)

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
# Load the dataset

df = pd.read_csv("StudentsPerformance.csv")

# Display the first 5 rows
df.head()
```

```python
# Display dataset information

print("Dataset shape:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()
```

```python
# Check the number of rows and columns

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
# Display column names

print("Dataset Columns:")
for column in df.columns:
    print(column)
```

```python
# Check dataset information

df.info()

print("\nMissing values in each column:")
print(df.isnull().sum())
```

```python
# Statistical summary of numerical columns

df.describe()
```

```python
# Summary of categorical columns

df.describe(include="object")
```

```python
# Check for missing values

print("Missing values in each column:")
print(df.isnull().sum())

print("\nTotal missing values:", df.isnull().sum().sum())
```

```python
# Check for duplicate rows

print("Number of duplicate rows:", df.duplicated().sum())
```

```python
# Create a copy of the original dataset

df_clean = df.copy()

# Remove duplicate rows
df_clean = df_clean.drop_duplicates()

# Check the cleaned dataset
print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", df_clean.shape)

print("\nMissing values after cleaning:")
print(df_clean.isnull().sum())
```

```python
# Calculate average scores

average_scores = df_clean[
    ["math score", "reading score", "writing score"]
].mean()

print("Average Exam Scores:")
print(average_scores)

# Create a bar chart
plt.figure(figsize=(8, 5))

average_scores.plot(kind="bar")

plt.title("Average Student Performance by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Score")
plt.xticks(rotation=0)
plt.tight_layout()

plt.savefig("average_student_scores.png")
plt.show()
```

```python
# 1. Math Score Distribution

plt.figure(figsize=(8, 5))

sns.histplot(data=df_clean, x="math score", bins=15, kde=True)

plt.title("Distribution of Math Scores")
plt.xlabel("Math Score")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.savefig("math_score_distribution.png")
plt.show()
```

```python
# 2. Score Comparison by Gender

plt.figure(figsize=(9, 5))

sns.boxplot(data=df_clean, x="gender", y="math score")

plt.title("Math Scores by Gender")
plt.xlabel("Gender")
plt.ylabel("Math Score")

plt.tight_layout()
plt.savefig("math_scores_by_gender.png")
plt.show()
```

```python
# 3. Test Preparation and Student Performance

score_columns = ["math score", "reading score", "writing score"]

prep_scores = df_clean.groupby("test preparation course")[score_columns].mean()

print(prep_scores)

prep_scores.plot(kind="bar", figsize=(9, 5))

plt.title("Average Exam Scores by Test Preparation Course")
plt.xlabel("Test Preparation Course")
plt.ylabel("Average Score")
plt.xticks(rotation=0)
plt.legend(title="Subject")
plt.tight_layout()

plt.savefig("test_preparation_analysis.png")
plt.show()
```

```python
# 4. Correlation Heatmap

score_columns = ["math score", "reading score", "writing score"]

correlation = df_clean[score_columns].corr()

plt.figure(figsize=(7, 5))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    fmt=".2f"
)

plt.title("Correlation Between Exam Scores")
plt.tight_layout()

plt.savefig("score_correlation_heatmap.png")
plt.show()
```

```python
# Calculate average exam scores

subjects = ["math score", "reading score", "writing score"]

print("AVERAGE SCORES")
print(df_clean[subjects].mean().round(2))

print("\nHIGHEST AVERAGE SCORE:")
print(df_clean[subjects].mean().idxmax())

print("\nLOWEST AVERAGE SCORE:")
print(df_clean[subjects].mean().idxmin())
```

```python
# Compare students by test preparation course

prep_analysis = df_clean.groupby("test preparation course")[subjects].mean().round(2)

print(prep_analysis)
```

```python
# Calculate correlations

correlation = df_clean[subjects].corr()

print("CORRELATION MATRIX:")
print(correlation.round(2))

print("\nCORRELATION BETWEEN READING AND WRITING:")
print(round(correlation.loc["reading score", "writing score"], 2))
```

```python
# Compare average scores by gender

gender_analysis = df_clean.groupby("gender")[subjects].mean().round(2)

print(gender_analysis)
```

```python
# Final Summary of Student Performance EDA

print("=" * 50)
print("STUDENT PERFORMANCE - EDA SUMMARY")
print("=" * 50)

print("\n1. DATASET OVERVIEW")
print("Total students:", len(df_clean))
print("Total columns:", len(df_clean.columns))

print("\n2. DATA QUALITY")
print("Missing values:", df_clean.isnull().sum().sum())
print("Duplicate rows:", df_clean.duplicated().sum())

print("\n3. AVERAGE SUBJECT SCORES")
print(df_clean[["math score", "reading score", "writing score"]].mean().round(2))

print("\n4. SCORES BY TEST PREPARATION")
print(df_clean.groupby("test preparation course")[
    ["math score", "reading score", "writing score"]
].mean().round(2))

print("\n5. CORRELATION MATRIX")
print(df_clean[["math score", "reading score", "writing score"]].corr().round(2))

print("\n6. SCORES BY GENDER")
print(df_clean.groupby("gender")[
    ["math score", "reading score", "writing score"]
].mean().round(2))

print("\nEDA SUMMARY COMPLETED!")
```
